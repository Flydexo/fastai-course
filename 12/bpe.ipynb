```python
from functools import reduce
```

```python
vocab = {}
token_list = []
last_token = 0
```

```python
data = open("/Users/matheoledevehat/Code/bpe/data.txt").read().lower()
```

```python
data
```

```python
for c in data:
    if c not in vocab:
        vocab[c] = last_token
        last_token+=1
        token_list.append(c)
```

```python
data_in_tokens = [c for c in data]
```

```python
while len(vocab.keys()) < 50:
    freqs = {}
    for i in range(len(data_in_tokens)-1):
        try:
            freqs[(data_in_tokens[i],data_in_tokens[i+1])]+= 1
        except:
            freqs[(data_in_tokens[i],data_in_tokens[i+1])] = 1
    a,b = reduce(lambda x,y: x if freqs[x] > freqs[y] else y, freqs.keys())
    most_popular_token = a+b
    vocab[most_popular_token] = last_token
    last_token+=1
    token_list.append(most_popular_token)
    i = 0
    new_data_in_tokens = []
    while i < len(data_in_tokens)-1:
        tok = data_in_tokens[i]
        if data_in_tokens[i]+data_in_tokens[i+1] in vocab:
            tok = data_in_tokens[i]+data_in_tokens[i+1]
            i+=1
        i+=1
        new_data_in_tokens.append(tok)
    data_in_tokens = new_data_in_tokens
    print(len(vocab.keys()))
```
