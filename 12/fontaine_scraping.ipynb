```python
import requests
import numpy as np
import pandas as pd
from functools import reduce
import re
from bs4 import BeautifulSoup
```

```python
url = "https://www.la-fontaine-ch-thierry.net/"
```

```python
recueil = requests.get(url+"fables.htm")
```

```python
soup = BeautifulSoup(recueil.text, 'html.parser')
```

```python
fable_urls = url+np.array([a.attrs['href'] for a in soup.select("td[width=\"738\"] p a")])
```

```python
def scrap_fable(url):
    try:
        return [a.text for a in BeautifulSoup(requests.get(url).text, 'html.parser').select("td td[width]")]
    except:
        print(url+" failed")
        return "scraping failed"
```

```python
fables = [scrap_fable(url) for url in fable_urls]
```

```python
fables = [f for f in fables if len(f) > 0]
```

```python
fables_texte = [fable[reduce(lambda x,y: x if len(fable[x]) > y else y, range(len(fable)))] for fable in fables]
```

```python
fables_texte = [reduce(lambda x,y: x+y, [txt for txt in fable if len(txt)>=100], "") for fable in fables]
```

```python
dump = reduce(lambda x,y: x+y, [re.sub("\((\d+|\*+)\)","",re.sub("\s+"," ",re.sub("\n+", "\n",fab).replace("\xa0+", "")).strip()) for fab in fables_texte])
```

```python
open("dump.txt", "w").write(dump)
```
